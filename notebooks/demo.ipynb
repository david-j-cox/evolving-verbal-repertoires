# Evolving Verbal Behavior — Demo

Phoneme-level simulation of Skinner's verbal operants. Words emerge from
shaping of random babbling via successive approximation — no a priori word
probabilities or pre-specified semantic structure.

Four operant systems (echoic, tact, mand, intraverbal) learn independently,
matching the functional independence described in Skinner's *Verbal Behavior*.

```python
import numpy as np
import matplotlib.pyplot as plt

import evb
from evb import SimulationConfig, run_simulation, analysis, visualization
```

## 1. Configuration

All parameters live in a single `SimulationConfig` dataclass.
Serialize to JSON for reproducibility.

```python
config = SimulationConfig(
    seed=42,
    n_steps=100_000,   # Use 500_000 for a full run
    log_every=1000,
)
print(config.to_json())
```

## 2. Run simulation

```python
%%time
log, learner, lexicon = run_simulation(config)
```

## 3. Dashboard

```python
phonology = evb.PhonologySystem()
fig = visualization.dashboard(log, learner, lexicon, phonology)
plt.show()
```

## 4. Analysis

```python
milestones = analysis.developmental_milestones(log, lexicon)
for key, val in milestones.items():
    print(f"{key}: {val}")
```

```python
emergence = analysis.word_emergence(log, lexicon, threshold=0.4)
emerged = {k: v for k, v in sorted(emergence.items(), key=lambda x: x[1] or float('inf')) if v is not None}
print(f"Words emerged (sim > 0.4): {len(emerged)} / {lexicon.n_words}")
for word, step in list(emerged.items())[:10]:
    print(f"  {word:10s} @ step {step:,}")
```

## 5. Individual plots

```python
fig = visualization.plot_shaping_trajectory(log, lexicon)
plt.show()
```

```python
fig = visualization.plot_confusion_heatmap(learner, lexicon, phonology, "echoic", n_samples=5)
plt.show()
```

```python
fig = visualization.plot_phoneme_accuracy(learner, lexicon, phonology, "echoic", n_samples=10)
plt.show()
```
